```python
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, accuracy_score
from sklearn.linear_model import LogisticRegression
from sklearn.svm import LinearSVC
from sklearn.preprocessing import StandardScaler
import os
import scipy.io as sio
```

```python
def filter_out_class_0(X, y):
    print("Filtering out Class 0 (Background)...")
    valid_pixels = y != 0  # Create a boolean mask where True means it's a real class
    X = X[valid_pixels]
    y = y[valid_pixels]

    return X, y
```

```python
# Load Data

main_path = os.path.join("..", "ds")

data_path = os.path.join(main_path, 'indiana_pines/indian_pines_corrected.mat')
labels_path = os.path.join(main_path, 'indiana_pines/indian_pines_gt.mat')

data = sio.loadmat(data_path)['indian_pines_corrected']
labels = sio.loadmat(labels_path)['indian_pines_gt']
```

```python
print("Reshaping data...")
# Flatten the 3D cube into a 2D matrix (Total Pixels x Bands)
X = data.reshape(-1, data.shape[-1])
# Flatten the labels to a 1D array
y = labels.reshape(-1)

# we keep all classes, including 0
print(f"Total samples being processed: {X.shape[0]}")

# --- THE FIX: FILTER OUT CLASS 0 ---
X_filtered, y_filtered = filter_out_class_0(X, y)
print(f"Total samples being processed: {X_filtered.shape[0]}")
```

# DECISION TREE

```python
# Split the dataset (70% training, 30% testing)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Initialize and train the single Decision Tree
print("Training Decision Tree (Including Class 0)...")
dt_classifier = DecisionTreeClassifier(random_state=42)
dt_classifier.fit(X_train, y_train)

# Predict on the test set
y_pred = dt_classifier.predict(X_test)

# Output the performance metrics
print("\n--- Decision Tree Performance ---")
print(f"Overall Accuracy: {accuracy_score(y_test, y_pred) * 100:.2f}%")
print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```

# LOGISTIC REGRESSION + LINEAR SVM

**1) Considering all the classes except class 0**

```python
# Split the dataset (70% training, 30% testing)
X_train, X_test, y_train, y_test = train_test_split(X_filtered, y_filtered, test_size=0.3, random_state=42)

# --- CRITICAL STEP FOR LINEAR MODELS: SCALING ---
print("Scaling data (Mean=0, Variance=1)...")
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Define the three linear models to test
models = {
    "Logistic Regression": LogisticRegression(max_iter=1000, random_state=42),
    "Linear SVM": LinearSVC(max_iter=1000, dual=False, random_state=42)
}

print("\n--- Testing Linear Separability ---")
for name, model in models.items():
    print(f"\n========================================")
    print(f"Training {name}...")
    model.fit(X_train_scaled, y_train)
    y_pred = model.predict(X_test_scaled)
    
    acc = accuracy_score(y_test, y_pred)
    
    print(f"Overall Accuracy: {acc * 100:.2f}%")
    print("Classification Report:")
    # Added zero_division=0 to prevent the warning block we saw earlier
    print(classification_report(y_test, y_pred, zero_division=0))
    print(f"========================================\n")
```

We are going to consider 2 specific subset that should be difficult to separate: **Corn** and **Soybeans**

In both cases we apply **Logistic Regression** and **Linear SVM**

```python
# Define our two test groups
crop_groups = {
    "Corn Tillage Group": [2, 3, 4],
    "Soybean Tillage Group": [10, 11, 12]
}

# Define the two linear models
models = {
    "Logistic Regression": LogisticRegression(max_iter=2000, random_state=42),
    "Linear SVM": LinearSVC(max_iter=2000, dual=False, random_state=42)
}

print("\n" + "="*50)
print(" ISOLATED LINEAR SEPARABILITY TEST: TILLAGE CLASSES")
print("="*50)

for group_name, target_classes in crop_groups.items():
    print(f"\n\n>>> Processing {group_name} {target_classes} <<<")
    
    # Filter dataset to ONLY include the current target classes
    mask = np.isin(y, target_classes)
    X_subset = X[mask]
    y_subset = y[mask]
    
    print(f"Total samples for {group_name}: {X_subset.shape[0]}")
    
    # Split the dataset (70% training, 30% testing)
    X_train, X_test, y_train, y_test = train_test_split(X_subset, y_subset, test_size=0.3, random_state=42)
    
    # Scale data
    scaler = StandardScaler()
    X_train_scaled = scaler.fit_transform(X_train)
    X_test_scaled = scaler.transform(X_test)
    
    # Test each model on this specific crop group
    for model_name, model in models.items():
        print(f"\n--- Training {model_name} on {group_name} ---")
        model.fit(X_train_scaled, y_train)
        y_pred = model.predict(X_test_scaled)
        
        acc = accuracy_score(y_test, y_pred)
        print(f"Accuracy: {acc * 100:.2f}%")
        print("Classification Report:")
        print(classification_report(y_test, y_pred, zero_division=0))

print("\n" + "="*50)
print(" EXPERIMENT COMPLETE")
print("="*50)
```
